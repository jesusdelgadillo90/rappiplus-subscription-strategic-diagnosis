```python
# PARTE 1: Totales del funnel
# ======================

query_totals = '''

WITH cte_first_visit AS (
   SELECT DISTINCT id_usuario
   FROM events
   WHERE nombre_evento = 'first_visit'
),

cte_add_to_cart AS(
SELECT DISTINCT id_usuario
FROM events
WHERE nombre_evento = 'add_to_cart'
),

cte_select_item AS(
SELECT DISTINCT id_usuario
FROM events
WHERE nombre_evento = 'select_item'
),

cte_begin_checkout AS(
SELECT DISTINCT id_usuario
FROM events
WHERE nombre_evento = 'begin_checkout'
),

cte_add_payment_info AS(
SELECT DISTINCT id_usuario
FROM events
WHERE nombre_evento = 'add_payment_info'
),

cte_purchase AS(
SELECT DISTINCT id_usuario
FROM events
WHERE nombre_evento = 'purchase'
)

SELECT
(SELECT COUNT(*) FROM cte_first_visit) AS first_visit_users,
(SELECT COUNT(*) FROM cte_add_to_cart) AS add_to_cart_users,
(SELECT COUNT(*) FROM cte_select_item) AS select_item_users,
(SELECT COUNT(*) FROM cte_begin_checkout) AS begin_checkout_users,
(SELECT COUNT(*) FROM cte_add_payment_info) AS add_payment_info_users,
(SELECT COUNT(*) FROM cte_purchase) AS purchase_users;

'''

totals = pd.read_sql(query_totals, con=engine)
totals
```

```python

# PARTE 2: Conversiones
# ======================

query_conversion = '''
WITH cte_first_visit AS (
   SELECT DISTINCT id_usuario
   FROM events
   WHERE nombre_evento = 'first_visit'
),

cte_add_to_cart AS(
SELECT DISTINCT id_usuario
FROM events
WHERE nombre_evento = 'add_to_cart'
),

cte_select_item AS(
SELECT DISTINCT id_usuario
FROM events
WHERE nombre_evento = 'select_item'
),

cte_begin_checkout AS(
SELECT DISTINCT id_usuario
FROM events
WHERE nombre_evento = 'begin_checkout'
),

cte_add_payment_info AS(
SELECT DISTINCT id_usuario
FROM events
WHERE nombre_evento = 'add_payment_info'
),

cte_purchase AS(
SELECT DISTINCT id_usuario
FROM events
WHERE nombre_evento = 'purchase'
)

SELECT
(SELECT COUNT(*) FROM cte_first_visit) AS first_visit_users,
(SELECT COUNT(*) FROM cte_add_to_cart) AS add_to_cart_users,
(SELECT COUNT(*) FROM cte_select_item) AS select_item_users,
(SELECT COUNT(*) FROM cte_begin_checkout) AS begin_checkout_users,
(SELECT COUNT(*) FROM cte_add_payment_info) AS add_payment_info_users,
(SELECT COUNT(*) FROM cte_purchase) AS purchase_users,

((SELECT COUNT(*) FROM cte_first_visit) - (SELECT COUNT(*) FROM cte_add_to_cart)) * 100.0
    / NULLIF((SELECT COUNT(*) FROM cte_first_visit), 0) AS dropoff_after_first_visit_pct,

    ((SELECT COUNT(*) FROM cte_add_to_cart) - (SELECT COUNT(*) FROM cte_select_item)) * 100.0
    / NULLIF((SELECT COUNT(*) FROM cte_add_to_cart), 0) AS dropoff_after_add_to_cart_pct,

    ((SELECT COUNT(*) FROM cte_select_item) - (SELECT COUNT(*) FROM cte_begin_checkout)) * 100.0
    / NULLIF((SELECT COUNT(*) FROM cte_select_item), 0) AS dropoff_after_select_item_pct,

    ((SELECT COUNT(*) FROM cte_begin_checkout) - (SELECT COUNT(*) FROM cte_add_payment_info)) * 100.0
    / NULLIF((SELECT COUNT(*) FROM cte_begin_checkout), 0) AS dropoff_after_begin_checkout_pct,

    ((SELECT COUNT(*) FROM cte_add_payment_info) - (SELECT COUNT(*) FROM cte_purchase)) * 100.0
    / NULLIF((SELECT COUNT(*) FROM cte_add_payment_info), 0) AS dropoff_after_add_payment_info_pct;

'''

conversion = pd.read_sql(query_conversion, con=engine)
conversion
```

## 🔹 Paso 4: Evaluar si los usuarios regresan (retención por cohortes)

**🎯 Objetivo:** Analizar la retención de usuarios para entender si regresan después de registrarse.

**Tablas**

- `users`
- `user_activity`

---
1. Se identifica la cohorte de cada usuario según el **mes de registro**.


2. Se calcula la retención semanal: cuántos usuarios **se mantienen activos** en cada semana desde su registro.
   - `retenido_w1`: usuarios activos en la semana 1  
   - `retenido_w2`: usuarios activos en la semana 2  
   - `retenido_w3`: usuarios activos en la semana 3  


3. Se calcula el porcentaje de retención para cada semana, dividiendo los usuarios retenidos entre los clientes iniciales de la cohorte:  
   - `semana_1`: porcentaje de usuarios retenidos en la semana 1  
   - `semana_2`: porcentaje de usuarios retenidos en la semana 2  
   - `semana_3`: porcentaje de usuarios retenidos en la semana 3  

Se revisa que la columna de fecha esté en formato correcto (`DATE`).  
Se realiza la conversión usando: `CAST(fecha_registro AS DATE)`

```python
# Explorar tabla users
# =========================
query_users = '''
SELECT *
FROM users;
'''
users = pd.read_sql(query_users, con=engine)
users.head(3)
```

```python
# Consulta para contar el número de fechas de registro por cliente

count_fecha_registro = '''

SELECT id_usuario,
COUNT(fecha_registro) AS conteo_fecha_registro
FROM users
GROUP BY id_usuario
HAVING count(fecha_registro) > 1
'''

fechas_registros = pd.read_sql(count_fecha_registro, con=engine)
fechas_registros
```

```python
# primera fecha de registro de los clientes

asignacion_cohorte = '''
SELECT id_usuario,
       DATE_TRUNC('week', CAST(MIN(fecha_registro) AS DATE)) AS cohorte_semanal
FROM users
GROUP BY id_usuario;
'''
fechas_cohortes = pd.read_sql(asignacion_cohorte, con=engine)
fechas_cohortes
```

```python
# Explorar tabla user_activity
# =========================
query_user_activity = '''
SELECT *
FROM user_activity;
'''
user_activity = pd.read_sql(query_user_activity, con=engine)
user_activity.head(3)
```

```python
# Retención por cohortes
# ======================
query_cohort_retention_completo = '''
WITH cohortes AS (
    -- Paso 1: Definir la cohorte mensual de cada usuario
    SELECT
        id_usuario,
        DATE_TRUNC('month', CAST(fecha_registro AS DATE)) AS cohorte_mensual
    FROM users
),
actividad_base AS (
    -- Paso 2: Unir con la actividad y calcular la semana usando dias_despues_registro
    SELECT
        c.id_usuario,
        c.cohorte_mensual,
        FLOOR(ua.dias_despues_registro / 7.0) AS week_number,
        ua.activo
    FROM cohortes c
    JOIN user_activity ua ON c.id_usuario = ua.id_usuario
),
total_usuarios_iniciales AS (
    -- Paso 3: Calcular el tamaño inicial (denominador) de cada cohorte mensual
    SELECT
        cohorte_mensual,
        COUNT(DISTINCT id_usuario) AS total_inicial
    FROM cohortes
    GROUP BY cohorte_mensual
),
retencion_semanal AS (
    -- Paso 4: Contar usuarios retenidos (activo = 1) para las semanas 1, 2 y 3
    SELECT
        cohorte_mensual,
        COUNT(DISTINCT CASE WHEN week_number = 1 AND activo = 1 THEN id_usuario END) AS retenido_w1,
        COUNT(DISTINCT CASE WHEN week_number = 2 AND activo = 1 THEN id_usuario END) AS retenido_w2,
        COUNT(DISTINCT CASE WHEN week_number = 3 AND activo = 1 THEN id_usuario END) AS retenido_w3
    FROM actividad_base
    GROUP BY cohorte_mensual
)
-- Paso 5: Unir todo y calcular los porcentajes de retención correspondientes
SELECT
    t.cohorte_mensual,
    t.total_inicial AS total_usuarios_iniciales,
    r.retenido_w1,
    r.retenido_w2,
    r.retenido_w3,
    ROUND(r.retenido_w1 * 100.0 / t.total_inicial, 2) AS semana_1_pct,
    ROUND(r.retenido_w2 * 100.0 / t.total_inicial, 2) AS semana_2_pct,
    ROUND(r.retenido_w3 * 100.0 / t.total_inicial, 2) AS semana_3_pct
FROM total_usuarios_iniciales t
JOIN retencion_semanal r ON t.cohorte_mensual = r.cohorte_mensual
ORDER BY t.cohorte_mensual;
'''

cohorte_mensual_final = pd.read_sql(query_cohort_retention_completo, con=engine)
cohorte_mensual_final
```

```python
# Verificación alternativa de retención basada en fechas reales
query_verificacion_fechas = '''
WITH actividad_calculada AS (
    SELECT
        u.id_usuario,
        DATE_TRUNC('month', CAST(u.fecha_registro AS DATE)) AS cohorte_mensual,
        -- Calculamos la diferencia en días y la convertimos a número de semana
        FLOOR((CAST(ua.fecha_actividad AS DATE) - CAST(u.fecha_registro AS DATE)) / 7.0) AS week_number,
        ua.activo
    FROM users AS u
    JOIN user_activity AS ua ON u.id_usuario = ua.id_usuario
    WHERE ua.fecha_actividad >= u.fecha_registro
),
total_inicial AS (
    SELECT
        DATE_TRUNC('month', CAST(fecha_registro AS DATE)) AS cohorte_mensual,
        COUNT(DISTINCT id_usuario) AS total_inicial
    FROM users
    GROUP BY cohorte_mensual
),
retencion_calculada AS (
    SELECT
        cohorte_mensual,
        COUNT(DISTINCT CASE WHEN week_number = 1 AND activo = 1 THEN id_usuario END) AS w1,
        COUNT(DISTINCT CASE WHEN week_number = 2 AND activo = 1 THEN id_usuario END) AS w2,
        COUNT(DISTINCT CASE WHEN week_number = 3 AND activo = 1 THEN id_usuario END) AS w3
    FROM actividad_calculada
    GROUP BY cohorte_mensual
)
SELECT
    t.cohorte_mensual,
    t.total_inicial,
    r.w1,
    r.w2,
    r.w3,
    ROUND(r.w1 * 100.0 / t.total_inicial, 2) AS w1_pct,
    ROUND(r.w2 * 100.0 / t.total_inicial, 2) AS w2_pct,
    ROUND(r.w3 * 100.0 / t.total_inicial, 2) AS w3_pct
FROM total_inicial t
JOIN retencion_calculada r ON t.cohorte_mensual = r.cohorte_mensual
ORDER BY t.cohorte_mensual;
'''

df_verificacion = pd.read_sql(query_verificacion_fechas, con=engine)
df_verificacion
```

```python
# Verificación individual de un usuario en user_activity
query_verificacion_usuario = '''
SELECT
    u.id_usuario,
    u.fecha_registro,
    ua.fecha_actividad,
    ua.dias_despues_registro,
    (CAST(ua.fecha_actividad AS DATE) - CAST(u.fecha_registro AS DATE)) AS dias_calculados_reales,
    ua.activo
FROM users AS u
JOIN user_activity AS ua ON u.id_usuario = ua.id_usuario
WHERE u.id_usuario = 'user_0'
ORDER BY ua.dias_despues_registro;
'''

df_usuario_check = pd.read_sql(query_verificacion_usuario, con=engine)
df_usuario_check
```

## 🔹 Paso 5: Validar si los cambios generan impacto (test estadístico)

🎯 **Objetivo:** Evaluar si la modificación en la UI del checkout impacta la **tasa de conversión de compra**.

---

1. **Analizar el dataset** `experiment_checkout_ui.csv` para identificar la métrica principal **conversion**.
   - La métrica **conversion** es 1 si el usuario completó la compra, 0 si no.    
2. **Plantear la hipótesis estadística**     
3. **Aplicar el test estadístico adecuado**
4. **Interpretar el resultado**  

---
Hipótesis estadística
   - **H₀ (Hipótesis nula):** La tasa de conversión es igual en ambas páginas. Cualquier diferencia se debe al azar del muestreo.
   - **H₁ (Hipótesis alternativa):** La tasa de conversión es diferente entre ambas páginas. Hay un factor que influye en la decisión de compra
   
**Test estadístico:** Z-Test

**Nivel de significancia alpha:** 0.05

```python
# tu código aquí
#==================================================================
#Import y configuración
#==================================================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from statsmodels.stats.proportion import proportions_ztest
```

```python
#=================================================================
#Carga Dataset
#=================================================================

df = pd.read_csv('https://practicum-content.s3.amazonaws.com/datasets/experiment_checkout_ui.csv')

print('Dimensiones del Dataset:', df.shape)
df.head()
```

```python
df.info()
```

```python
# Duplicados por usuario
duplicated_users = df.duplicated(subset=['id_usuario']).sum()
print('Usiarios Duplicados:', duplicated_users)
```

```python
categorical_cols = ['variante', 'dispositivo', 'pais']

for col in categorical_cols:
    print('\n---', col, '---')
    display(df[col].value_counts())
```

# Totales

```python
#============================================================
# Contar número de exitos, cuantos usuarios, de cada variante, si compraron
#============================================================

conversiones = df.groupby('variante')['convirtio'].sum()
display(conversiones)
```

```python
#============================================================
# Contar número total de usuarios de cada variante
#============================================================

totales = df.groupby('variante')['convirtio'].count()
display(totales)
```

```python
#============================================================
# Organizamos los datos en listas: una con el número de éxitos
#============================================================

exitos = [conversiones['control'], conversiones['tratamiento']]
display(exitos)
```

```python
#============================================================
# Organizamos los datos en listas: con el total de observaciones por grupo
#============================================================

observaciones = [totales['control'], totales['tratamiento']]
display(observaciones)
```

# Calculo de prueba Z

```python
z_stat, p_value = proportions_ztest(exitos, observaciones)

print(f'Estadístico z: {z_stat}')
print(f'Valor p: {p_value}')
```

```python
#========================================================
# Umbral de significancia
#=========================================================
alpha = 0.05

if p_value < alpha:
    print('Rechazamos la Hipótesis Nula: hay evidencia de una diferencia.')
else:
    print('No rechazamos la Hipótesis Nula: no hay evidencia suficiente de una diferencia.')
```

### 📝 Conclusión del Paso 5: Evaluación de Impacto

Tras analizar los resultados del experimento A/B en el archivo `experiment_checkout_ui.csv` mediante una prueba Z para dos proporciones, se concluye lo siguiente:

1. **Planteamiento de Hipótesis:**
   - **$H_0$ (Hipótesis Nula):** La tasa de conversión es igual entre el grupo de control y la variante de tratamiento. Cualquier diferencia observada se debe al azar del muestreo.
   - **$H_1$ (Hipótesis Alternativa):** La tasa de conversión difiere entre ambas variantes.

2. **Resultado del Test:**
   - El estadístico Z obtenido es **-0.8133** y el **p-value es de 0.41606**.

3. **Interpretación y Decisión de Negocio:**
   - Dado que el p-value ($0.41606$) es sustancialmente mayor que el nivel de significancia estándar ($\alpha = 0.05$), **no se rechaza la hipótesis nula ($H_0$)**.
   - **Recomendación:** No existe evidencia estadística suficiente para afirmar que la modificación en la UI del checkout haya incrementado (o disminuido) de forma real la tasa de conversión. Por lo tanto, se recomienda evaluar los costos de implementación frente a la falta de impacto observado antes de desplegar el cambio de manera definitiva
